```python
import random
nums = list()
for i in range (10):
    num = random.randint (0,10)
    nums.append (num)

nums_tpl = tuple (nums)
print (nums_tpl)

tpl1 = tuple ([num for num in list(nums_tpl) if num % 2 == 0 or num % 3 == 0])
print (tpl1)

lst1 = list()
for num in list(nums_tpl):
    if num % 2 == 0 and num % 3 == 0:
        lst1.append (num)
tpl2 = tuple (lst1)
print (tpl2) 

for num in nums_tpl: 
    if num % 2 == 0 and num % 3 == 0:
        print (num)     
```
