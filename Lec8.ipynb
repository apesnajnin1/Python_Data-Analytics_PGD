```python
phrase = 'Meet me at backyard'
encrypt = str()
for char in phrase: 
    encrypt += chr(ord(char) + 3)
print (encrypt)

decrypt = str()
for char in encrypt:
    decrypt += chr(ord(char) - 3)
print (decrypt)
```

```python
names = ['Alice', 'Bob', 'David']
ages = [25, 31, 23]

for name, age in zip(names, ages):
    print (name, age)
```
