```python
#SampleQUestion3: extract and display all unique teachers from the Teacher column.
#uteachers = df['Teacher'].unique()
uteachers = set(df['Teacher'])
#uteachers = df['Teacher'].nunique()
print(uteachers)

```

```python
fteachers = df['Teacher'].value_counts()
print(fteachers)
```

```python
fteachers1 = df.groupby('Teacher').size().reset_index(name='Student Count')
print(fteachers1)
```

```python
#SampleQuestion4: find all students enrolled in the course BUS 102.
std_bus102 = df[df['Course Code'] == 'BUS 102']
print(std_bus102)
```

```python
#SampleQuestion5:Calculate the number of students in each course.
students = df.groupby('Course Code').size().reset_index(name='Student Count')
print(students)
```

```python
#SampleQuestion6:Count how many students applied for the exam.
exam_applied = df['Exam Type'].value_counts()
print(exam_applied)
```

```python
student_no_of_makeup = df['Student Id'].value_counts()
print(student_no_of_makeup)
```

```python
#SampleQuestion7:Find all students who applied on or after a specific date, e.g., 10-05-2024.
target_date = pd.to_datetime('10-05-2024', dayfirst = True)
print(target_date)
df['Application Date'] = pd.to_datetime(df['Applied Date'], dayfirst = True)
last_applied_students = df[df['Application Date'] >= target_date]
print(last_applied_students)

```
