```python
n = int(input('Put a number: '))

sum = 0
i = 1
while i <= n:
    sum += i
    i += 1
print (f'Summation: {sum}')
```

```python
n = int(input('Put a number: '))

fact = 1
i = 1
while i <= n:
    fact *= i
    i += 1
print (f'Factorial: {fact}')
```

```python
n = int (input ('Put a number: '))
sum = 0
while n > 0:
    digit = n % 10
    sum += digit
    n = int (n/10)
print (f'Summation: {sum}')
```
