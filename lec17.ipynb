```python
import pandas as pd
data = [10, 20, 30, 40]
label = ['a', 'b', 'c', 'd']
series = pd.Series(data, index=label)
print (series)
print (series.iloc[2])
print (series['c'])
```

```python
data = {'Name': ['Alice', 'Bob', 'Charlie'], 'Age':[25, 30, 35]}
print (data)
df = pd.DataFrame (data)
print (df)
print (df['Age'])
print (df.iloc[1])
print (df.loc[2, 'Age'])
```

```python
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df['C'] = [7, 8, 9]
print (df)
df = df.drop('A', axis=1)
print (df)
```

```python
df = pd.DataFrame({'Category': ['A', 'A', 'B', 'B'], 'Value': [10, 15, 20, 25]})
grouped = df.groupby('Category').median()
print(grouped)
```

```python
import pandas as pd
df = pd.DataFrame({'Category': ['A', 'A', 'B', 'B'], 'Value': [10, 15, 20, 25]})
grouped = df.groupby('Category').sum()
print(grouped)
```

```python
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
print(df.aggregate('max')) # Apply sum function to all columns
```

```python
import numpy as np
df = pd.DataFrame({'A': [1, np.nan, 3], 'B': [4, 5, np.nan]})
df = df.fillna(5)
print (df)
list_ = df['A']
print (sum(list_))
```

```python
dates = pd.date_range(start='20230101', periods=6, freq='B')
df = pd.DataFrame(np.random.randn(6, 4), index=dates, columns=['humidity', 'temp', 'zzz', 'hhh'])
print(df)
```

```python
df1 = pd.DataFrame({'key': ['A', 'B'], 'value': [1, 2]})
df2 = pd.DataFrame({'key': ['A', 'C'], 'value': [3, 4]})
merged = pd.merge(df1, df2, on='key', how='inner')
print(merged)
```

```python
df1 = pd.DataFrame([[1, 2], [3, 4]], columns=['A', 'B'])
df2 = pd.DataFrame([[5, 6], [7, 8]], columns=['A', 'B'])
result = pd.concat([df1, df2], axis=0) # Concatenate along rows
print(result)
```
