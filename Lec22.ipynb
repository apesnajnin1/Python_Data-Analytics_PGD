```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0, 2, 4, 6])
y = np.array([0, 100, 150, 250])
plt.plot (x,y,'v-.k')

x = np.array([0, 2, 3, 6])
y = np.array([5, 60, 120, 240])
plt.plot (x,y,'o-r')

plt.xlabel ('Overs')
plt.ylabel ('Runs')
plt.xticks ([0, 1, 2, 3, 4, 5, 6])
plt.yticks ([0, 50, 100, 150, 200, 250])
plt.grid()
plt.savefig ('text.svg')
plt.savefig ('text.eps')
plt.savefig ('text.pdf')
plt.show ()
```
