```python
def Factorial (n):
    fact = 1
    for i in range (1, n+1):
        fact *= i
    return fact

def SumOfSeries (fn, ln):
    sum = 0
    for i in range (fn, ln+1):
        sum += Factorial(i)
    return sum

print (f'Sum:{SumOfSeries(1, 5)}')
print (f'Sum:{SumOfSeries(2, 8)}')
print (f'Sum:{SumOfSeries(1, 7)}')
```

```python
import math

def Prime (n):
    prime = True
    for i in range (2, int(math.sqrt(n))+1):
        if n % i == 0:
            prime = False
            break
    return prime

def SumOfSeries (fn, ln):
    sum = 0
    for i in range (fn, ln+1):
        if Prime (i) == True:
            sum += i
    return sum

print (f'Summation: {SumOfSeries(2, 100)}')
```
