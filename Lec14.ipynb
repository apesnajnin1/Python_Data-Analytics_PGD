```python
words = list()
for i in range (5): 
    word = input (f'Put word no {i+1}: ') 
    words.append(word)
print (words)

for word in words:
    if word[-1] == 'n':
        print (word)
```

```python
import random

NUM = 5

nlist = [random.randint(0, 10) for i in range(NUM)]
print(nlist) 

num = int(input(f'Put a number: '))
idx = nlist.index(num)
if idx != None:
    num1 = int(input(f'Put the new number: '))
    nlist[idx] = num1
print (nlist)
```
