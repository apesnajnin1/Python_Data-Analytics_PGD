```python
attempt = 0
while True:
    password = input ('Password: ')
    i = u = d = s = 0

    while i < len(password):
        if password[i].isalpha():
            if password[i].isupper():
                u += 1
            i += 1
            continue
        elif password[i].isdigit():
            d += 1
            i += 1
            continue
        else:
            s += 1
            i += 1

    if u >= 1  and d >= 1 and s >= 1 and len(password) >= 8:
        break

    attempt += 1
    if attempt == 3:
        break

if attempt < 3:
    print ('New password is set')
else:
    print ('New password is not set')
```
