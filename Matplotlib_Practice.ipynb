```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([0, 6])
y = np.array([0, 250])
plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, 'o')
plt.show()
```

```python
plt.xticks([-2. -1, 0, 1, 2, 3])
plt.yticks([-2, -1, 0, 1, 2, 3])
plt.plot(x,y)
plt.show()
```

```python
y = np.array([3, 8, 5, 6, 8])
plt.plot(y)
plt.show()
```

```python
x = np.array([0, 6])
y = np.array([0, 250])
plt.plot(x, y, 'v-k')

x = np.array([0, 6])
y = np.array([0, 280])
plt.plot(x, y, 'o-')

plt.show()

```
