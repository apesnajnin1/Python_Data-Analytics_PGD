```python
#HW3a: Counting word Frequencies
list_of_words = ['apple', 'banana', 'strawberry', 'apple', 'pineapple', 'banana']
wdcounts = dict()
for w in list_of_words:
    wdcounts[w] = wdcounts.get(w, 0) + 1
print(wdcounts)
```

```python
#HW3b: What if there are duplicate values? If the word count example,
#more than one word appears 2 times in the text...How do we deal with that?
list_of_words = ['apple', 'banana', 'strawberry', 'apple', 'pineapple', 'banana']
wdcnt = dict()
for w in list_of_words:
    wdcnt[w] = wdcnt.get(w, 0) + 1
print(wdcnt)
wdcnt_reverse = dict()
for w in wdcnt:
  c = wdcnt[w]
  if c in wdcnt_reverse:
    wdcnt_reverse[c].append(w)
  else:
    wdcnt_reverse[c] = [w]
print (wdcnt_reverse)
```

```python
#HW5: Reverse a Dictionary: Given a dictionary {1: 'a', 2:'b', 3:'c'},
# reverse its key-value pairs to get {'a':1, 'b':2, 'c':3}.
original_dict = {1: 'a', 2: 'b', 3: 'c'}
reverse_dict = dict()
for w in original_dict:
  c = original_dict[w]
  if c in reverse_dict:
    reverse_dict[c].append(w)
  else:
    reverse_dict[c] = [w]
print (reverse_dict)
```

```python
#HW6: Word Frequency Counter: Write a program to count the frequency of
# each word in the sentence 'the quick brown fox jumps over the lazy dog'.

sentence = 'the quick brown fox jumps over the lazy dog'
word_count = dict()
for w in sentence.split():
    word_count[w] = word_count.get(w, 0) + 1
print(word_count)
```

```python
#HW7: Nested Dictionary Manipulation: Create a nested dictionary to store student
#information:
# students = {
#   'Alice':{'math': 85, 'science': 92},
#    'Bob':{'math': 78, 'science': 89},
#    'Charlie':{'math': 95, 'science': 85}
#     }
# Calculate the average score of each student and print it.

students = {
    'Alice': {'math': 85, 'science': 92},
    'Bob': {'math': 78, 'science': 89},
    'Charlie': {'math': 95, 'science': 85}
}
# Iterate through the student names (this uses the dictionary's __iter__() method)
for name in students:
    # Access the inner dictionary containing subjects and scores
    scores_dict = students[name]
    print(scores_dict)

    # Extract the numerical scores using the .values() method
    scores = scores_dict.values()
    print(scores)

    # Calculate the average by dividing the sum of scores by the count of subjects
    average = sum(scores) / len(scores)

    # Print the result formatted to two decimal places
    print(f"{name}'s Average Score: {average:.2f}")
```

```python
#HW8: Merge and handle Conflicts: Write a function to merge two dictionaries.
#If a key exists in both, sum their values:
# dict1 = {'a': 5, 'b':10, 'c':15}
#dict2 = {'b': 20, 'c': 25, 'd': 30}
#
# Expected Output:{'a': 5, 'b': 30, 'c': 40, 'd': 30}

dict1 = {'a': 5, 'b':10, 'c':15}
dict2 = {'b': 20, 'c': 25, 'd': 30}
merged_dict = dict1.copy()
for key, value in dict2.items():
    if key in merged_dict:
        merged_dict[key] += value
    else:
        merged_dict[key] = value
print(merged_dict)
```

```python
#HW9: Group By Key Prefix: Given a dictionary with keys in the format 'prefix_suffix',
# group values by their prefixes:
# data = {'a_1':10, 'a_2': 20`, 'b_1': 15, 'b_2': 25, 'c_1': 30}
# Expected Output:{'a': [10, 20], 'b': [15, 25], 'c': [30]}.

data = {'a_1':10, 'a_2': 20, 'b_1': 15, 'b_2': 25, 'c_1': 30}
grouped_dict = dict()
for key, value in data.items():
    prefix = key.split('_')[0]
    print(prefix)
    if prefix in grouped_dict:
        grouped_dict[prefix].append(value)
    else:
        grouped_dict[prefix] = [value]

print(grouped_dict)
```

```python
#Exercise 2: Write a Python Program to Get 5 Numbers from User and Create a
#Dictionary which Displays Square of Every Numbers as Value.

# Initialize an empty dictionary
squares_dict = {}

print("Please enter 5 numbers:")

# Loop 5 times to get input from the user
for i in range(5):

    try:
        num = float(input(f"Enter number {i+1}: "))

        squares_dict[num] = num ** 2

    except ValueError:
        print("Invalid input. Please enter a numeric value.")

        i -= 1

# Display the resulting dictionary
print("\nDictionary of squares of Numbers:")
print(squares_dict.values())

```

```python
#Problem 1 : Write a Python Program to Create a Dictionary from a List
# which Displays Length of Each Word as Value.

# List of words
words = ["apple", "banana", "orange", "date", "jackfruit", "mango"]

# Dictionary comprehension

word_lengths = {word: len(word) for word in words}

# Printing the result
print(word_lengths)
```

```python
# Problem 2: Write a Python Program to Create a Dictionary
# from a List which Displays Length of those Word which
# Contain Greater than 5 Character as Value.

# List of words
words_list = ["sparrow", "crow", "parrot", "peacock", "cuckoo", "magpie", "kingfisher"]

# Dictionary comprehension with a condition
# Syntax: {key: value for item in iterable if condition}
word_lengths = {word: len(word) for word in words_list if len(word) > 5}

# Display the result
print("Original List:", words_list)
print("Resulting Dictionary:", word_lengths)
```
