```python
fh = int(input("Put fahrenheit: "))
print (fh)

celcius = (fh - 32) * 5 / 9
print ('Celcius: {:.2f}'.format(celcius))

if celcius < 0:
    print ('Freezing')
    if celcius < -30:
        print ('Extremely cold!')
else:
    print ('Not freezing')
    if celcius >= 40:
        print ('Extremely hot!')

```
