```python
def Largest ():
    num1 = 23
    num2 = 45
    num3 = 12

    largest = None    
    if num1 > num2 and num1 > num3:
        largest = num1
    elif num2 > num1 and num2 > num3:
        largest = num2
    else:
        largest = num3

    print (f'Largest: {largest}')

Largest ()
```

```python
def Largest ():
    lst = [23, 45, 12, 78, 345, 78, 12, 56]
    largest = lst[0]
    for i in range (1, len(lst)):
        if lst[i] > largest:
            largest = lst[i]
    print (f'Largest: {largest}')
Largest ()
```

```python
def Factorial ():
    num = int (input('Put a number: '))
    fact = 1
    for i in range (1, num + 1):
        fact *= i
    print (f'factorial of {num}: {fact}')
Factorial ()
Factorial ()
Factorial ()
```
