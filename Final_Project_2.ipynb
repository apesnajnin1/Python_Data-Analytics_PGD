```python
# 1. Load your data
from google.colab import files
uploaded = files.upload()
import pandas as pd
df = pd.read_csv('japan_suicide_info.csv')
print(df)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

# 1. Clean Data
# Removes hidden spaces in column names (e.g., 'Sex ' -> 'Sex')
df.columns = df.columns.str.strip()

# Convert 'Sex' to numbers (Male=0, Female=1) if it's currently text
if df['Sex'].dtype == 'object':
    df['Sex_Numeric'] = df['Sex'].map({'Male': 0, 'Female': 1, 'Unknown': 2})
else:
    df['Sex_Numeric'] = df['Sex']

# Drop missing values to ensure the statistical test runs
df_clean = df.dropna(subset=['Sex_Numeric', 'Death rate per 100 000 population'])

# 2. Correlation Matrix & Heatmap
# Only include numeric columns for the matrix
corr_matrix = df_clean.select_dtypes(include=['number']).corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='RdBu_r', center=0, fmt=".2f")
plt.title('Correlation Heatmap: Mortality Metrics')
plt.show()

# 3. Statistical Significance
# Using Point-Biserial as it is optimized for Binary (Sex) vs Continuous (Rate) data
corr, p_value = stats.pointbiserialr(
    df_clean['Sex_Numeric'],
    df_clean['Death rate per 100 000 population']
)

print("-" * 30)
print(f"Analysis Results:")
print(f"Pearson Correlation: {corr:.4f}")
print(f"P-Value:             {p_value:.4f}")
print("-" * 30)

if p_value < 0.05:
    print("STATUS: Statistically Significant")
    print("There is a reliable relationship between Sex and Death Rate.")
else:
    print("STATUS: Not Significant")
    print("The relationship may be due to random chance.")
```

```python
# 1. Load your data
from google.colab import files
uploaded = files.upload()
import pandas as pd
df = pd.read_csv('data_Japan.csv')
print(df)
```

```python

df_clean = df.dropna()
# Split features (X) and Target (y)
X = df_clean.drop([ '0 Year','1-4 Years', '5-9 Years', 'Unknown Age',
                   'No of Suicides',
                    'Percentage of cause-specific deaths out of total deaths',
                    'Death rate per 100 000 population', 'Sex_Numeric'], axis=1)
print('X after drop columns:\n', X)

Y = df_clean[['Sex_Numeric', 'No of Suicides', 'Percentage of cause-specific deaths out of total deaths', 'Death rate per 100 000 population']]
print('\n\nThe Target Columns:\n',Y)
# Convert categorical text into numbers (One-Hot Encoding)
X = pd.get_dummies(X)
print('\n\nTHe Converted X:\n', X)
```

```python
import pandas as pd
from sklearn.feature_selection import VarianceThreshold, SelectKBest, f_regression

# 1. Remove Low-Variance Features
# Drops features where more than 99% of the values are the same
selector_var = VarianceThreshold(threshold=0.01)
X_reduced_array = selector_var.fit_transform(X)

# Convert back to DataFrame to keep track of column names
X_reduced = pd.DataFrame(X_reduced_array, columns=X.columns[selector_var.get_support()])

print(f"Features removed by variance: {list(X.columns[~selector_var.get_support()])}")

# 2. Select Top 5 Features
# f_regression is used because our targets (suicide rates) are continuous numbers.
selector_k = SelectKBest(score_func=f_regression, k=5)

# We use Y.iloc[:, 0] (Sex Numeric) as the reference for importance
X_new_array = selector_k.fit_transform(X_reduced, Y.iloc[:, 0])
print(X_new_array)

# 3. Get and display results
selected_features = X_reduced.columns[selector_k.get_support()]
print(f"Top 5 Selected Features: {list(selected_features)}")

# Final reduced DataFrame
X_final = pd.DataFrame(X_new_array, columns=selected_features)
print(X_final)
```

```python
from sklearn.model_selection import train_test_split

#Y = df_clean[['Sex_Numeric', 'No of Suicides',
#                'Percentage of cause-specific deaths out of total deaths',
#                'Death rate per 100 000 population']]

# 1. Define your target (Y)
Y = df_clean['Sex_Numeric']
#Y = df_clean['No of Suicides']
Y1 = df_clean['Percentage of cause-specific deaths out of total deaths']
Y2 = df_clean['Death rate per 100 000 population']

# 2. Split the data
# X_final contains the above top 5 features
# test_size=0.2 reserves 20% of the data for testing
X_train, X_test, Y_train, Y_test = train_test_split(
    X_final,
    Y,
    test_size=0.2,
    random_state=42
)

# 3. Verify the split
print(f"Training set size: {X_train.shape[0]} rows")
print(f"Testing set size:  {X_test.shape[0]} rows")
print(f"Features used:     {list(X_train.columns)}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. Initialize the model
lr_model = LinearRegression()

# 2. Train the model using the selected features and multi-output target
# We use X_new (or X_final) and the y from our previous train/test split
lr_model.fit(X_train[selected_features], Y_train)

# 3. Make predictions on the test set
Y_pred = lr_model.predict(X_test[selected_features])

# 4. Evaluate the model
print("--- Model Performance ---")
print(f"Overall R² Score: {r2_score(Y_test, Y_pred):.4f}")
print(f"Overall MSE:      {mean_squared_error(Y_test, Y_pred):.4f}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

# 1. Scale the features (Essential for Logistic Regression)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 2. Initialize and Train the model
log_reg = LogisticRegression(random_state=42)
log_reg.fit(X_train_scaled, Y_train)

# 3. Make Predictions
Y_pred = log_reg.predict(X_test_scaled)

# 4. Evaluation
print(f"Accuracy Score: {accuracy_score(Y_test, Y_pred):.4f}")
print("\n Confusion Matrix:")
print(confusion_matrix(Y_test, Y_pred))
print("\n Detailed Classification Report:")
print(classification_report(Y_test, Y_pred))
```
