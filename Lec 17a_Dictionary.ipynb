```python
tt = dict()
for i in range (1, 6):
    tt[i] = i * i

for key in tt.keys():
    print (f'key: {key}')
```

```python
digital_items = dict()

for i in range (2): 
    name, price = input ('Put name and price: ').split(' ')
    price = float(price)

    digital_items[name] = price

print (f'Summation: {sum(digital_items.values())}')
```

```python
di = {'mouse': 500, 'keyboard': 700, 'monitor': 7000, 'pendrive': 1000, 'cable': 500}
fdi = {key:value + value*0.15 for key, value in di.items()}
print (fdi)

fdi1 = {key:value + value*0.15 for key, value in di.items() if value >= 1000}
print (fdi1)

fdi2 = {key:value + value*0.15 if value >= 1000 else value + value*0.10 for key, value in di.items()}
print (fdi2)

fdi3 = {key:value for key, value in di.items() if value >= 1000 if value <= 4000}
print (fdi3)

```
